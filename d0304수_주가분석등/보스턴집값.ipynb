```python
# 범죄율과 주택 가격의 산점도 분석
plt.scatter(df['CRIM'],df['MEDV'],alpha=0.5)
plt.title('범죄율과 주택 가격의 산점도 분석')
plt.xlabel('범죄율')
plt.ylabel('주택가격')
plt.show()
```

```python
df['CRIM_G'] = pd.cut(df['CRIM'], bins=10)
grouped = df.groupby('CRIM')['MEDV'].mean()
grouped.plot(label='범죄율')
plt.show()
```

```python
# 일산화질소 농도와 주택 가격의 산점도 분석

plt.scatter(df['NOX'],df['MEDV'],alpha=0.5)
plt.title('일산화질소농도와 주택 가격의 산점도 분석')
plt.xlabel('일산화질소')
plt.ylabel('주택가격')
plt.show()
```
