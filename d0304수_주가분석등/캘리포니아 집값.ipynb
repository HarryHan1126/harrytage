- 가져온 데이터를 데이터프레임으로 만들기

- california 데이터 확인

- 가져온 데이터를 판다스 데이터프레임으로 만들기

```python
df = pd.DataFrame(data, columns=california.feature_names)
df['Price'] = target
df.index = df.index + 1
df
```

```python
df.corr()
```

- 동네 중간 소득(MedInc)'과 '집값(Price)'의 상관계수 계산 ( corr함수 )

```python
df['MedInc'].corr(df['Price'])
```

- 산점도(Scatter) 그리기 
- x축은 소득(MedInc), y축은 집값(Price)
- alpha=0.1은 점을 투명하게 만들어서 겹치는 부분을 진하게 보여줌

```python
plt.scatter(df['MedInc'], df['Price'], alpha=0.1, color='g', label='corr')
plt.xlabel('MedInc')
plt.ylabel('Price')
plt.legend()
plt.show()
```
