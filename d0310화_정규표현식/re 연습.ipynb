- text에서 이메일 형식을 찾아 리스트로 추출

```python
text = "제 이메일은 test1234@gmail.com 이고, 회사 이메일은 work_email@company.co.kr 입니다."
```

- text에서 한글, 영문, 띄어쓰기만 남기고 나머지 모두 제거

- text에서 url 형식만 리스트로 추출

```python
tex = """
검색은 주로 구글(https://www.google.com)을 사용하고, 
뉴스는 네이버(http://naver.com)에서 봅니다. 
가끔 보안이 안 된 http://old-site.org 도 접속하죠.
"""
```

```python
p = re.compile(r'\w+://(?:\w+\.)*.+\.\w+')
m2 = p.findall(tex)
print(m2)
```

```python
p = re.compile(r'\w+://(\w+\.)*.+\.\w+')
m2 = p.findall(tex)
print(m2)
```
