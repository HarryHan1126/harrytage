```python
import cv2
cap = cv2.VideoCapture('test.mp4')

fps = cap.get(cv2.CAP_PROP_FPS) #약 30
delay = round(1000 / (fps*2))
width = round(cap.get(cv2.CAP_PROP_FRAME_WIDTH)) #540
height = round(cap.get(cv2.CAP_PROP_FRAME_HEIGHT)) #960
fourcc = cv2.VideoWriter_fourcc(*'DIVX') #코덱 설정
out = cv2.VideoWriter('video_out.avi', fourcc, fps*2, (height,width))

while cap.isOpened():
    ret, frame = cap.read()
    
    if not ret :
        break
        
    rotate90 = cv2.rotate(frame, cv2.ROTATE_90_COUNTERCLOCKWISE)
    
    cv2.imshow('elphant', rotate90)
    out.write(rotate90)
    
    if cv2.waitKey(delay) == ord('q') :
        print('사용자에 의해 종료되었습니다.')
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
img = cv2.imread('img.jpg')

rotate90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE) # 시계방향 90도 회전
rotate180 = cv2.rotate(img, cv2.ROTATE_180)  # 180도 회전
rotate270 = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE) # 시계반대방향 90도

cv2.imshow('img', img)
cv2.imshow('rotate90', rotate90)
cv2.imshow('rotate180', rotate180)
cv2.imshow('rotate270', rotate270)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
