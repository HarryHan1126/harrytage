```python
import cv2
import mediapipe as mp

cap = cv2.VideoCapture('early.mp4')

mp_face_detection = mp.solutions.face_detection
mp_drawing = mp.solutions.drawing_utils

fps = cap.get(cv2.CAP_PROP_FPS)
width = round(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = round(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'DIVX')
out = cv2.VideoWriter('early_out.avi', fourcc, fps, (width, height))

COLOR = (255, 0, 0)   # 빨간색(BGR)

with mp_face_detection.FaceDetection(
    model_selection=0, min_detection_confidence=0.5) as face_detection:

    while cap.isOpened():
        success, image = cap.read()
        if not success:
            break

        # 성능 향상용
        image.flags.writeable = False
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        results = face_detection.process(image)

        # 다시 그릴 수 있게 변경
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

        # 현재 프레임 크기 구하기
        h, w, _ = image.shape

        if results.detections:
            for detection in results.detections:
                bbox = detection.location_data.relative_bounding_box
                x = int(bbox.xmin * w)
                y = int(bbox.ymin * h)

                # print("------------------------")
                # print(f"얼굴 위치: x={x}, y={y}")

                # keypoint 출력 + 양쪽 눈에 원 그리기
                for i, keypoint in enumerate(detection.location_data.relative_keypoints):
                    px = int(keypoint.x * w)
                    py = int(keypoint.y * h)
                    # print(f"keypoint {i}: ({px}, {py})")

                    # 0: 오른쪽 눈, 1: 왼쪽 눈
                    if i == 0 or i == 1:
                        cv2.circle(image, (px, py), 15, COLOR, cv2.FILLED)

                mp_drawing.draw_detection(image, detection)

        cv2.imshow('MediaPipe Face Detection', image)
        out.write(image)

        if cv2.waitKey(50) == ord('q'):
            break

cap.release()
out.release()
cv2.destroyAllWindows()
```
