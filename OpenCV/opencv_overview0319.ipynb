1. 이미지 읽기 옵션
  - cv2.IMREAD_COLOR : 컬러 이미지, 투명 영역은 무시 (디폴트)
  - cv2.IMREAD_GRAYSCALE : 흑백 이미지
  - cv2.IMREAD_UNCHANGED : 투명 영역까지 포함 

```python
import cv2
img_color = cv2.imread('img.jpg', cv2.IMREAD_COLOR)
img_gray = cv2.imread('img.jpg', cv2.IMREAD_GRAYSCALE)
img_unchanged = cv2.imread('img.jpg', cv2.IMREAD_UNCHANGED)

cv2.imshow('img_color', img_color)
cv2.imshow('img_gray', img_gray)
cv2.imshow('img_unchanged', img_unchanged)

if cv2.waitKey(0) == ord('A') :
    print('사용자가 A를 눌렀음.') #cv2.waitKey(0) 0 이라고 하면 무한대로 창을 띄움, 50이면 50밀리초, 1000이면 1초 이런식으로 화면을 띄움
cv2.destroyAllWindows()
```

2. 동영상 보여주기
   - 동영상 파일 보여주기

```python
import cv2
cap = cv2.VideoCapture('video.mp4')

while cap.isOpened():
    ret, frame = cap.read() #read는 두가지 정보, 리턴 됫는지 성공여부, 해당 이미지를 불러옴
    if not ret : 
        print('더 이상 가져올 이미지가 없습니다.')
        break
    cv2.imshow('video', frame)

    if cv2.waitKey(50) == ord('q') :
        print('사용자에 의해 종료되었습니다.')
        break

cap.release()
cv2.destroyAllWindows()
```

- 직선 그리기
  - cv2.LINE_4 : 상하좌우 4가지 방향으로만 연결된 선
  - cv2.LINE_8 : 상하좌우 + 대각선을 포함 8가지 방향으로 연결된 선 (디폴트)
  - cv2.LINE_AA : 부드러운 선 (anti-aliasing)

- cv2.line( 그릴 위치(도화지), 시작점(x,y), 끝점, 색깔, 두께, 선 종류 )

```python
import cv2
import numpy as np

img = np.zeros((480, 640, 3), dtype=np.uint8)

COLOR = (0, 0, 255)
THIC = 3

cv2.line( img, (50,50), (500, 400), COLOR, THIC, cv2.LINE_AA)
cv2.line( img, (400,50), (50, 500), COLOR, THIC, cv2.LINE_AA)

cv2.imshow('line', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

- 원 그리기
  - cv2.circle( 그릴 위치, 원의 중심점, 반지름, 색깔, 두께, 선종류)

```python
import cv2
import numpy as np

img = np.zeros((480, 640, 3), dtype=np.uint8)
img[:] = (255,255,255)

COLOR = (0, 0, 255)
THIC = 10
RADIUS = 100

# cv2.circle( img, (300,200), RADIUS, COLOR, THIC, cv2.LINE_AA)
cv2.circle( img, (300,200), RADIUS, COLOR, cv2.FILLED, cv2.LINE_AA)

cv2.imshow('circle', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

- 사각형 그리기
  - cv2.rectangle (그릴 위치, 왼쪽위 좌표, 오른쪽 아래 좌표, 두께)

```python
import cv2
import numpy as np

img = np.zeros((480, 640, 3), dtype=np.uint8)

COLOR = (200, 145, 255)
THIC = 10

# cv2.circle( img, (300,200), RADIUS, COLOR, THIC, cv2.LINE_AA)
cv2.rectangle( img, (100,100), (400,400), COLOR, THIC)
cv2.rectangle( img, (125,125), (375,375), COLOR, THIC)
cv2.rectangle( img, (150,150), (350,350), COLOR, THIC)
cv2.rectangle( img, (175,175), (325,325), COLOR, THIC)
cv2.rectangle( img, (200,200), (300,300), COLOR, THIC)
cv2.rectangle( img, (225,225), (275,275), COLOR, cv2.FILLED)

cv2.imshow('rectangle', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

- 다각형 그리기
  - cv2.polylines( 그릴 위치, 그릴 좌표들, 닫힘여부(T/F), 색깔, 두께, 선종류 )
    : 비어있는 다각형 그릴 때
  - cv2.fillPoly( 그릴 위치, 그릴 좌표들, 색깔, 선 종류 )
    : 채워져있는 다각형 그릴 때

```python
import cv2
import numpy as np

img = np.zeros((480, 640, 3), dtype=np.uint8)

COLOR = (200, 145, 255)
THIC = 3

pt1 = np.array([[[100, 100], [200, 100], [100, 200]], [[200, 100], [300, 100], [300, 200]]])
pt2 = np.array([[[100, 300], [200, 300], [100, 400]], [[200, 300], [300, 300], [300, 400]]])

cv2.polylines( img, pt1, True, COLOR, THIC, cv2.LINE_AA )
cv2.fillPoly( img, pt2, COLOR, cv2.LINE_AA )

# 3차원으로 만들었기 때문에 pt1만 넣어도 3차원 array 로 인식 
# pt1 = np.array([[[100, 100], [200, 100], [100, 200]],[[200, 100], [300, 100], [300, 200]]])
# cv2.polylines( img, pt1, True, COLOR, THIC, cv2.LINE_AA )

cv2.imshow('polylines', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# 검은 도화지
img = np.zeros((600, 800, 3), dtype=np.uint8)

# 기본 삼각형 2개
pt1 = np.array([
    [[100, 100], [200, 100], [100, 200]],
    [[200, 100], [300, 100], [300, 200]]
], dtype=np.int32)

# 여러 줄로 반복 배치
for row in range(4):      # 세로 4줄
    for col in range(3):  # 가로 3칸
        moved = pt1 + [col * 220, row * 120]

        # 첫 번째 삼각형 채우기
        cv2.fillPoly(img, [moved[0]], (0, 255, 255))   # 노랑
        # 두 번째 삼각형 채우기
        cv2.fillPoly(img, [moved[1]], (255, 100, 0))   # 파랑 계열

# 화면 출력
cv2.imshow('pattern', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
