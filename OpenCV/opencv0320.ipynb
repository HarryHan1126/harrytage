```python
import cv2
import numpy as np

img = np.zeros((600, 600, 3), dtype=np.uint8)

color1 = (0, 200, 0)
color2 = (0, 0, 255)
THIC = 3


pt1 = np.array([[300, 100], [200, 250], [400, 250]])
pt2 = np.array([[100, 500], [150, 350], [200, 425], [250, 350], [300, 500]])
# pt2 = np.array([[[100, 300], [200, 300], [100, 400]], [[200, 300], [300, 300], [300, 400]]])

cv2.polylines( img, [pt1], True, color1, THIC, cv2.LINE_AA )
cv2.fillPoly( img, [pt2], color2, cv2.LINE_AA )


cv2.imshow('polylines', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
result = cv2.imwrite('test.png', img)
print(result)  # True면 성공
```
