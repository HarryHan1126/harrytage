## 이미지 자르기
- 이미지 잘라서 새로운 창에 표시

```python
import cv2
img = cv2.imread('img.jpg')
crop = img[215:485, 48:330] #세로 기준값, 가로 기준값
# img.shape => (640, 427, 3)

cv2.imshow('img', img)
cv2.imshow('crop', crop)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 이미지 블러(흐리게)
 - 커널 사이즈 변화에 따른 블러 (가우시안 블러)

```python
img = cv2.imread('img.jpg')

# 커널 사이즈 => (3,3) (5,5) (7,7)
kenel_33 = cv2.GaussianBlur(img, (3,3), 0)
kenel_55 = cv2.GaussianBlur(img, (5,5), 0)
kenel_77 = cv2.GaussianBlur(img, (7,7), 0)

cv2.imshow('img',img)
cv2.imshow('kenel_33',kenel_33)
cv2.imshow('kenel_55',kenel_55)
cv2.imshow('kenel_77',kenel_77)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

 - 표준 편차에 따른 블러

```python
img = cv2.imread('img.jpg')

# 커널 사이즈 => (3,3) (5,5) (7,7)
sigma_1 = cv2.GaussianBlur(img, (0,0), 1) # sigmaX 가우시안 블러의 X축 방향의 표준편차
sigma_2 = cv2.GaussianBlur(img, (0,0), 2)
sigma_3 = cv2.GaussianBlur(img, (0,0), 3)

cv2.imshow('img',img)
cv2.imshow('sigma_1',sigma_1)
cv2.imshow('sigma_2',sigma_2)
cv2.imshow('sigma_3',sigma_3)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 이미지 다림질 효과 (원근감)

```python
import numpy as np

img = cv2.imread('news.jpg')

width, height = 640, 200

src = np.array([[508,381],[1020,371],[1120,587],[457,600]], dtype=np.float32) #입력위치값
dst = np.array([[0,0],[width,0],[width,height],[0,height]], dtype=np.float32) #출력위치값
#4개 꼭지점은 시계방향으로 좌표값 입력 

matrix = cv2.getPerspectiveTransform(src, dst)
res = cv2.warpPerspective(img, matrix, (width,height))


cv2.imshow('news', img)
cv2.imshow('res', res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import math

point4 = []
img = cv2.imread('card.jpg')
COLOR = (0, 0, 255)
width, height = 320, 450

def mouse_handler(event, x, y, flag, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        point4.append((x, y))
        print(f'x좌표:{x}, y좌표:{y}')
        if len(point4) == 4:
            show_result()
        
        for point in point4:
            cv2.circle(img, point, 10, COLOR, cv2.FILLED)
            cv2.imshow('img', img)
            
def distance(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)
    
def show_result():
    print('결과 이미지 출력 성공')

    dists = []
    for i in range(4):
        d = distance(point4[i], point4[(i+1) % 4])
        dists.append(d)

    #가로,세로 구하기 
    height = int(max(dists))
    width = int(min(dists))

    src = np.array(point4, dtype=np.float32)
    dst = np.array([[0,0],[width,0],[width,height],[0,height]], dtype=np.float32)
    
    matrix = cv2.getPerspectiveTransform(src, dst)
    res = cv2.warpPerspective(img, matrix, (width,height))
    cv2.imshow('result', res)
    
cv2.namedWindow('img')
cv2.setMouseCallback('img', mouse_handler)

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
