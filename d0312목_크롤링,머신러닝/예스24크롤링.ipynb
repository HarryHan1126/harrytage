```python
import requests
from bs4 import BeautifulSoup
import lxml
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
# 한페이지 24, top 100 , 제목,저자,가격,리뷰,판매지수,책제목에 어느 단어가 많이 들어가는
```

```python
#모바일IT 120권
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

total_count = 0 
all_books = []

for page in range(1,6):
    url = f'https://www.yes24.com/product/category/daybestseller?pageNumber={page}&pageSize=24&categoryNumber=001001003&type=day'
    driver.get(url)
    time.sleep(2)
    
    print(f"IT/모바일분야 '{page}'번째 페이지 도서 정보 수집을 시작합니다.\n" + "="*40)
    
    books = driver.find_elements(By.CSS_SELECTOR, 'div.itemUnit')

    time.sleep(1)
    
    for i,book in enumerate (books,1):
        title = book.find_element(By.CSS_SELECTOR, 'a.gd_name').text
        author = book.find_element(By.CSS_SELECTOR, 'span.authPub.info_auth').text
        cost = book.find_element(By.CSS_SELECTOR, 'strong.txt_num em.yes_b').text
        review_elem = book.find_elements(By.CSS_SELECTOR, 'span.rating_grade em.yes_b')
        review = review_elem[0].text if review_elem else '리뷰 없음'
        sales_volume = book.find_element(By.CSS_SELECTOR, 'span.saleNum').text

        print(f"{i}. {title}")
        print(f" *저자 : {author}")
        print(f" *가격 : {cost}")
        print(f" *리뷰 : {review}점")
        print(f" *{sales_volume}")

        total_count += 1

        d = {
            'title' : title,
            'author' : author,
            'cost' : cost,
            'review' : review,
            'sales_volume' : sales_volume
        }

        all_books.append(d)
        
driver.quit()
print("="*40 + "\n" f"총 {total_count}건 수집이 완료되었습니다!")
```

```python
df_120 = pd.DataFrame(all_books)
df_120.index.name = 'No'
df_120.index += 1 
df_120.to_csv('Books120.csv',encoding='utf-8-sig')
df_120.head()        
```

```python
#주간베스트 30권, +할인율, 태그까지 
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)
total_count = 0 
weekly_best = []

for page in range(1,3):
    url = f'https://www.yes24.com/product/category/weekbestseller?categoryNumber=001&pageNumber={page}&pageSize=24&type=week'
    driver.get(url)
    time.sleep(2)
    
    print(f"주간베스트 '{page}'번째 페이지 도서 정보 수집을 시작합니다.\n" + "="*40)
    
    books = driver.find_elements(By.CSS_SELECTOR, 'div.itemUnit')

    time.sleep(1)
    
    for i,book in enumerate (books,1):
        if total_count >= 30:
            break
        
        title = book.find_element(By.CSS_SELECTOR, 'a.gd_name').text
        author = book.find_element(By.CSS_SELECTOR, 'span.authPub.info_auth').text
        cost = book.find_element(By.CSS_SELECTOR, 'strong.txt_num em.yes_b').text
        
        sale_elem = book.find_elements(By.CSS_SELECTOR, 'span.txt_sale')
        sale = sale_elem[0].text if sale_elem else '할인 없음'
        
        review_elem = book.find_elements(By.CSS_SELECTOR, 'span.rating_grade em.yes_b')
        review = review_elem[0].text if review_elem else '리뷰 없음'
        sales_volume = book.find_element(By.CSS_SELECTOR, 'span.saleNum').text
        
        tag_elem = book.find_elements(By.CSS_SELECTOR, 'span.tag')
        tag = ', '.join([t.text for t in tag_elem]) if tag_elem else '태그 없음'
        
        print(f"{i}. {title} | 저자 : {author} | 가격 : {cost} | 할인 : {sale}%")
        print(f" 리뷰 : {review}점 | {sales_volume}")
        print(f" 태그 : {tag}")

        total_count += 1

        b = {
            'title' : title,
            'author' : author,
            'cost' : cost,
            'review' : review,
            'sales_volume' : sales_volume,
            'sale' : sale,
            'tags' : tag
        }

        weekly_best.append(b)
    
    if total_count >= 30:
        break
driver.quit()
print("="*40 + "\n" f"총 {total_count}건 수집이 완료되었습니다!")
```

```python
df_30 = pd.DataFrame(weekly_best)
df_30.index.name = 'No'
df_30.index += 1 
df_30.to_csv('weekly30.csv',encoding='utf-8-sig')
df_30.head()    
```
